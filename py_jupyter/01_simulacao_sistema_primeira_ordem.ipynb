```python
# Experimento 01 - Simulação de um sistema dinâmico

import numpy as np
import matplotlib.pyplot as plt

# Tempo de simulação
tempo = np.linspace(0, 10, 100)

# Parâmetros do sistema
K = 1
tau = 1

# Resposta de um sistema de primeira ordem
saida = K * (1 - np.exp(-tempo / tau))

# Gráfico
plt.plot(tempo, saida)
plt.xlabel("Tempo (s)")
plt.ylabel("Saída do sistema")
plt.title("Resposta de um sistema de primeira ordem")
plt.grid(True)
plt.show()
```
